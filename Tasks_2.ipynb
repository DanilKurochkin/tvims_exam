```python
from math import factorial
from collections import Counter

def count_unique_permutations(word):
    # Подсчет количества каждой буквы
    letter_counts = Counter(word)
    
    # Вычисляем факториал длины слова
    total_permutations = factorial(len(word))
    
    # Делим на факториалы количества каждой буквы
    for count in letter_counts.values():
        total_permutations //= factorial(count)
    
    return total_permutations
```

```python
word = "AAA"
print(f"Количество различных слов: {count_unique_permutations(word)}")
```

```python
word = "Данил"
print(f"Количество различных слов: {count_unique_permutations(word)}")
```
