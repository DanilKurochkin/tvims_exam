Квантили и процентные точки случайной величины. Привести пример расчета 10% проц.точки на небольшом датасете.

```python
from scipy.stats import skew, kurtosis
import numpy as np
import random

random.seed(42)

data = [random.randint(1, 100) for _ in range(10)]
print(f"Данные: {data}")

skewness = skew(data)
print(f"Асимметри: {np.round(skewness, 2)}")

kurt = kurtosis(data)
print(f"Эксцесс: {np.round(kurt, 2)}")
```
