```python
import numpy as np
import random
```

```python
random.seed(42)

data = [random.randint(-23, 24) for _ in range(100)]
```

```python
# Описательная статистика
mean_value = np.mean(data)           # Среднее
median_value = np.median(data)       # Медиана
std_dev = np.std(data)               # Стандартное отклонение
min_value = np.min(data)             # Минимум
max_value = np.max(data)             # Максимум
iqr = np.percentile(data, 75) - np.percentile(data, 25)  # Межквартильный размах

# Вывод результатов
print(f"Среднее значение: {np.round(mean_value, 3)}")
print(f"Медиана: {np.round(median_value, 3)}")
print(f"Стандартное отклонение: {np.round(std_dev, 3)}")
print(f"Минимум: {np.round(min_value, 3)}")
print(f"Максимум: {np.round(max_value, 3)}")
print(f"Межквартильный размах (IQR): {np.round(iqr, 3)}")
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_data = pd.DataFrame(data, columns=['random_value'])

fig, ax = plt.subplots(figsize=(11, 6), dpi=150)
plt.title("Boxplot for random value")
sns.boxplot(df_data, x='random_value')

ax.axvline(median_value, color='red', label=f"median: {median_value}")
ax.axvline(mean_value, color='green', label=f"mean: {mean_value}")

plt.legend()
```
